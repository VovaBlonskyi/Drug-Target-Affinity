```python
def Mr(smiles):
    mol = Chem.MolFromSmiles(smiles)
    weight = Descriptors.ExactMolWt(mol)
    return weight
```

```python
def logP(smiles):
    mol = Chem.MolFromSmiles(smiles)
    logp = Crippen.MolLogP(mol)
    return logp
```

```python
def HBonds(smiles):
    mol = Chem.MolFromSmiles(smiles)
    Hdonors = Lipinski.NumHDonors(mol)
    Hacceptors = Lipinski.NumHAcceptors(mol)
    return {
        "HDonors":Hdonors,
        "HAcceptors":HAceptors
    }
```

```python
def TPSA(smiles):
    mol = Chem.MolFromSmiles(smiles)
    Tpsa = MolSurf.TPSA(mol)
    return Tpsa
```

```python
def aromaticCycles(smiles):
    mol = Chem.MolFromSmiles(smiles)
    heteroAromatic = Lipinski.NumAromaticHeterocycles(mol)
    carboAromatic = Lipinski.NumAromaticCarbocycles(mol)
    return {
        "aromaticHeterocycles":heteroAromatic,
        "aromaticCarbocycles":carboAromatic
    }
```

```python
def formalCharge(smiles):
    
    mol = Chem.MolFromSmiles(smiles)
    overall_charge = 0
    for i in mol.GetAtoms():
        overall_charge += i.GetFormalCharge()

    return overall_charge
```

```python
def hybridization(smiles):
    
    mol = Chem.MolFromSmiles(smiles)
    SP3 = 0
    SP2 = 0
    SP = 0
    
    for i in mol.GetAtoms():
        hybridization = i.GetHybridization()
        if hybridization == Chem.rdchem.HybridizationType.SP3:
            SP3 +=1
        elif hybridization == Chem.rdchem.HybridizationType.SP2:
            SP2 +=1
        elif hybridization == Chem.rdchem.HybridizationType.SP:
            SP +=1

    return {
        'SP3' : SP3,
        'SP2' : SP2,
        'SP' : SP
    }
```

```python
def bondFeatures(smiles):
    mol = Chem.MolFromSmiles(smiles)

    single = 0
    double = 0
    triple = 0
    aromatic = 0
    conjugated = 0
    total_bonds = mol.GetNumBonds()
    ringBonds = 0
    
    for bond in mol.GetBonds():
        if bond.IsInRing():
            ringBonds +=1
        btype = bond.GetBondType()
        if btype == Chem.rdchem.BondType.SINGLE:
            single += 1
        elif btype == Chem.rdchem.BondType.DOUBLE:
            double += 1
        elif btype == Chem.rdchem.BondType.TRIPLE:
            triple += 1
        elif btype == Chem.rdchem.BondType.AROMATIC:
            aromatic += 1
        
        if bond.GetIsConjugated():
            conjugated += 1


    return {
        "single_bonds": single,
        "double_bonds": double,
        "triple_bonds": triple,
        "aromatic_bonds": aromatic,
        "rotatable_bonds": Lipinski.NumRotatableBonds(mol),
        "conjugated_bonds": conjugated,
        "conjugation_ratio": conjugated / total_bonds if total_bonds else 0,
        "bonds_in_ring": ringBonds
    }
```

```python
import rdkit
import pandas as pd
from pyopsin import PyOpsin
from rdkit import Chem
from rdkit.Chem import Draw
pd.set_option('display.max_colwidth', None)
```

```python
pyopsin = PyOpsin()
```

```python
smiles = pyopsin.to_smiles('(2S)-2-amino-3-(1H-indol-3-yl)propanoic acid')
save = ''
for i in smiles:
    save = i
print(smiles)
```

```python
draw = Chem.MolFromSmiles(save)
dr = Chem.MolFromSmiles('COc1cc(OC)cc(c1)C1Cc2cnc(cc2N=C1NC(=O)NC(C)(C)C)N(CCCOCc1ccccc1)C(C)=O |c:20|')
Draw.MolToImage(draw)
Draw.MolToImage(dr)
```

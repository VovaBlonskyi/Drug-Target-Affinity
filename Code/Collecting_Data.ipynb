```python
import pandas as pd
import sqlite3
import numpy as np
import json
```

```python
pd.set_option('display.max_colwidth', None)
```

```python
binding = pd.read_csv(r'C:\Users\blons\code\Chem_AI\Drug-Target\Databases\BindingDB_All.tsv', sep='\t', usecols = ['Target Source Organism According to Curator or DataSource', 'Target Name'])
binding
```

```python
binding['Target Name'] = binding['Target Name'].fillna('').astype(str)
binding['Target Source Organism According to Curator or DataSource'] = (
    binding['Target Source Organism According to Curator or DataSource']
    .fillna('')
    .astype(str)
)
```

```python
Target_and_Species = {}
```

```python
for index, row in binding.iterrows():
    combo = row['Target Name'] + ' : ' + row['Target Source Organism According to Curator or DataSource']
    if combo in Target_and_Species:
        Target_and_Species[combo] += 1
    else:
        Target_and_Species[combo] = 1
```

```python
Target_and_Species
```

```python
with open('Target_and_Species.json', 'w') as fp:
    json.dump(Target_and_Species, fp)
```

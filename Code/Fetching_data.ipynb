```python
import pandas as pd
import os
import numpy as np
```

```python
pd.set_option('display.max_colwidth', None)

pd.set_option("display.max_columns", None)
```

```python
binding = pd.read_csv(r'C:\Users\blons\code\Chem_AI\Drug-Target\Databases\BindingDB_All.tsv', sep='\t', usecols = ['BindingDB Target Chain Sequence 1', 'Ligand SMILES','Ki (nM)'], on_bad_lines='skip', low_memory=False)
```

```python
chunk_size = 100000

trial_num = int(input("Enter the trial number: "))

os.makedirs("chunks", exist_ok=True)

for i in range(0, len(binding), chunk_size):

    chunk = binding.iloc[i:i+chunk_size]

    filename = f"chunks/chunk_{i//chunk_size}_{trial_num}.csv"

    chunk.to_csv(filename, index=False)
```

```python
for i in range( round( len( binding.index ) / chunk_size ) + 1 ):  
    
    chunk = pd.read_csv(f"chunks/chunk_{i}_{trial_num}.csv")
    chunk = chunk.dropna(subset = ["Ki (nM)"])
    chunk.to_csv(f"chunks/chunk_{i}_{trial_num}.csv")
```

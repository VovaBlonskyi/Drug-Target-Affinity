```python
from propy.PyPro import GetProDes
from Bio.SeqUtils.ProtParam import ProteinAnalysis
```

```python
def AAC(sequence):
    des = GetProDes(sequence)
    return des.GetAAComp()
```

```python
def DPC(sequence):
    des = GetProDes(sequence)
    return des.GetDPComp()
```

```python
def CTD(sequence):
    des = GetProDes(sequence)
    return des.GetCTD()
```

```python
def PAAC(sequence):
    des = GetProDes(sequence)
    return des.GetPAAC(lamda=10, weight=0.05)
```

```python
def APAAC(sequence):
    des = GetProDes(sequence)
    return des.GetAPAAC(lamda=10, weight=0.05)
```

```python
def MB(sequence):
    des = GetProDes(sequence)
    return des.GetMoreauBrotoAuto()
```

```python
def Moran(sequence):
    des = GetProDes(sequence)
    return des.GetMoranAuto()
```

```python
def Geary(sequence):
    des = GetProDes(sequence)
    return des.GetGearyAuto()
```

```python
def Mr(sequence):
    analysis = ProteinAnalysis(sequence)
    return analysis.molecular_weight()
```

```python
def Aromaticity(sequence):
    analysis = ProteinAnalysis(sequence)
    return analysis.aromaticity()
```

```python
def IP(sequence):
    analysis = ProteinAnalysis(sequence)
    return analysis.isoelectric_point()
```

```python
def II(sequence):
    analysis = ProteinAnalysis(sequence)
    return analysis.instability_index()
```

```python
def GRAVY(sequence):
    analysis = ProteinAnalysis(sequence)
    return analysis.gravy()
```

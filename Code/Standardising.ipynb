```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_parquet("training_data.parquet")
```

```python
nunique = df.nunique()
    
columns = nunique[nunique > 2].index.tolist()
    
for i in columns:

    if i == 'pKi':
        continue
    
    mean = df[i].mean()
    std = df[i].std()
    
    df[i] = (df[i] - mean) / std
```

```python
df.to_parquet("standardised_training_data.parquet", index = None)
```
